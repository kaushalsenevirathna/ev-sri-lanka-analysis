## Missing / Invalid `manufacture_year` values

- 4 rows had missing `manufacture_year` in the original data.
- 8 additional rows had implausible years (e.g. 202, 1320, 2055, 2106) outside a realistic
  range and were set to NaN, since the true intended year could not be reliably inferred.
- Decision: these 12 rows (0.18% of the dataset) are kept in the cleaned dataset rather than
  dropped, since vehicle_category, make, district, and count are still valid and useful for
  non-time-based analyses. They will be excluded only from analyses that specifically require
  manufacture_year (e.g. time trend charts), using `df[df['manufacture_year'].notna()]`.

```python
print(df['make'].nunique())
print(sorted(df['make'].unique())[:40])
```

```python
df['make'] = df['make'].str.strip().str.upper()
print(df['make'].nunique())
```

```python
print(sorted(df['make'].unique()))
```

```python
from difflib import get_close_matches

makes = sorted(df['make'].unique())
for m in makes:
    matches = get_close_matches(m, makes, n=3, cutoff=0.8)
    matches = [x for x in matches if x != m]
    if matches:
        print(m, '->', matches)
```

```python
print(df[df['make'] == 'MG4'][['make','model']])
print(df[df['make'] == 'NISSAN MT'][['make','model']])
```

```python
make_corrections = {
    'CITROAN': 'CITROEN',
    'DONG FENG': 'DONGFENG',
    'VOLKSWAGON': 'VOLKSWAGEN',
    'YADIA': 'YADEA',
}
df['make'] = df['make'].replace(make_corrections)
```

```python
df['manufacture_year'] = df['manufacture_year'].astype('Int64')
print(df['manufacture_year'].dtype)
print(df['manufacture_year'].head(10))
```

## Data Cleaning Summary

- Loaded raw DMT Excel data (6,536 rows), skipping the header row and dropping empty extra columns.
- Renamed columns to snake_case: vehicle_category, make, model, manufacture_year, district, count.
- Fixed a district typo: "invalid vavuniaya" -> VAVUNIYA.
- Standardized `make` values: stripped whitespace, uppercased for consistency.
- Set 8 implausible manufacture_year values (e.g. 202, 1320, 2055, 2106) to missing, since the
  true year could not be reliably inferred. Combined with 4 originally missing years, 12 rows
  (0.18%) now have unknown manufacture_year but are retained for non-time-based analysis.
- Used fuzzy string matching (difflib) to identify likely typo duplicates in `make`, then manually
  verified each candidate against its associated `model` values before merging. Merged: CITROAN->CITROEN,
  DONG FENG->DONGFENG, VOLKSWAGON->VOLKSWAGEN, YADIA->YADEA, MG4->MG, NISSAN MT->NISSAN.
  Left ambiguous look-alikes (e.g. ELECTRA/ELECTRICA, TAILG/TAILING/TAING) unmerged due to
  insufficient evidence they represent the same brand.
- Replaced a single placeholder make value "-" with "UNKNOWN".
- Converted manufacture_year to a nullable integer type (Int64).
- Saved cleaned dataset to data/processed/ev_cleaned.csv.
